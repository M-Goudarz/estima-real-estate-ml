```python
from pathlib import Path

import numpy as np
import pandas as pd

PROJECT_ROOT = Path.cwd().parent
DATA_PATH = PROJECT_ROOT / "data" / "processed" / "tehran_houses_clean.csv"

df = pd.read_csv(DATA_PATH)

print("Dataset shape:", df.shape)
display(df.head())
```

```python
TARGET = "Price"

NUMERIC_FEATURES = [
    "Area",
    "Room",
]

BOOLEAN_FEATURES = [
    "Parking",
    "Warehouse",
    "Elevator",
]

CATEGORICAL_FEATURES = [
    "Address",
]

FEATURES = NUMERIC_FEATURES + BOOLEAN_FEATURES + CATEGORICAL_FEATURES

print("Target:", TARGET)
print("Numeric features:", NUMERIC_FEATURES)
print("Boolean features:", BOOLEAN_FEATURES)
print("Categorical features:", CATEGORICAL_FEATURES)

print("\nNumber of input features:", len(FEATURES))
```

```python
# Analyze categorical features
for feature in CATEGORICAL_FEATURES:
    print(f"Feature: {feature}")
    print(f"Number of unique categories: {df[feature].nunique()}")
    print(f"Missing values: {df[feature].isna().sum()}")
    print("\nTop 10 categories:")
    display(df[feature].value_counts().head(10))
    print("-" * 50)
```

```python
address_counts = df["Address"].value_counts()

print("Number of categories:", len(address_counts))
print("Categories with fewer than 5 samples:", (address_counts < 5).sum())
print("Categories with fewer than 10 samples:", (address_counts < 10).sum())

print("\nLeast frequent categories:")
display(address_counts.sort_values().head(15))
```

## Feature Engineering Summary

The feature engineering stage focused on analyzing the existing input features and determining their suitability for machine learning.

### 1. Feature Definition
- Defined `Price` as the target variable.
- Identified six input features:
  - Numerical: `Area`, `Room`
  - Boolean: `Parking`, `Warehouse`, `Elevator`
  - Categorical: `Address`

### 2. Categorical Feature Analysis
- Identified 193 unique categories in the `Address` feature.
- Found that several neighborhoods have very few observations.
- Decided to retain all categories to avoid losing potentially useful geographical information.

### 3. Feature Selection Decisions
- Retained all six input features.
- Excluded `Price(USD)` due to target leakage.
- Avoided creating features derived directly from `Price`.

### Conclusion

No additional engineered features were introduced at this stage. The existing features were retained for model development, with categorical encoding and numerical scaling to be handled within the machine learning preprocessing pipeline.
